```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../output.log")
```

```python
import matplotlib.colors

for col in [
    #"count",
    #"ewma_2_perc",
    #"ewma_5_perc",
    #"ewma_20_perc",
    #"ewma_100_perc",
    #"global_count_since_top1_percent_ewma5",
    #"group_-3_mean_ewma5_perc",
    #"group_-2_mean_ewma5_perc",
    #"group_-1_mean_ewma5_perc",
    #"group_-3_mean_ewma5_perc",
    #"group_0_mean_ewma5_perc",
    #"group_1_mean_ewma5_perc",
    #"group_2_mean_ewma5_perc",
    #"group_3_mean_ewma5_perc",
    "age_count_total",
    "ewma_100_malloc_calls",
    "age",
    "discounted_reward_90",
    "discounted_reward_95",
    "discounted_reward_99",
]:
    pivot_df = df.pivot(index="page", columns="step", values=col)
    sns.heatmap(pivot_df, cmap="YlOrRd", annot=False, norm=matplotlib.colors.LogNorm()).set(title=col)
    plt.show()
```
